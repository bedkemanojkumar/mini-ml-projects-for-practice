```python
#Import Required Libraries
import re
from difflib import get_close_matches
```

```python
#Text Preprocessing
def preprocess_text(text):

    # Convert to lowercase
    text = text.lower()

    # Remove punctuation
    text = re.sub(r'[^\w\s]', '', text)

    return text
```

```python
#Preprocess FAQ Questions
processed_questions = [preprocess_text(q) for q in faq_questions]

vectorizer = TfidfVectorizer(stop_words='english')

faq_vectors = vectorizer.fit_transform(processed_questions)
```

```python
conversation_history = []

while True:

    user_query = input("\nAsk your question (type 'exit' to quit): ")

    # Exit condition
    if user_query.lower() == "exit":
        print("AI Agent: Goodbye!")
        break

    # Store query in memory
    conversation_history.append(user_query)

    # Preprocess query
    processed_query = preprocess_text(user_query)

    # Typo correction
    vocabulary = vectorizer.get_feature_names_out()

    corrected_words = []

    for word in processed_query.split():

        matches = get_close_matches(word, vocabulary, n=1, cutoff=0.7)

        if matches:
            corrected_words.append(matches[0])
        else:
            corrected_words.append(word)

    corrected_query = " ".join(corrected_words)

    print("Corrected Query:", corrected_query)

    # Convert query into vector
    user_vector = vectorizer.transform([corrected_query])

    # Compute similarity
    similarity_scores = cosine_similarity(user_vector, faq_vectors)

    # Best match
    best_match_index = similarity_scores.argmax()

    max_similarity = similarity_scores.max()

    print("Similarity Score:", max_similarity)

    # Response logic
    if max_similarity > 0.5:
        print("AI Agent:", faq_answers[best_match_index])
    else:
        print("AI Agent: Sorry, I don't understand the question.")
```

```python
# For viewing conversation,chat history
print("\nConversation History:")

for i, query in enumerate(conversation_history, 1):
    print(f"{i}. {query}")
```

```python
# Show Top 3 Similar Questions
top_indices = similarity_scores.argsort()[0][-3:][::-1]

print("\nTop Matching Questions:")

for idx in top_indices:
    print(f"- {faq_questions[idx]}")
```

```python
#Convert Your Logic into Function
def chatbot(user_query):

    # Preprocess query
    processed_query = preprocess_text(user_query)

    # Get vocabulary
    vocabulary = vectorizer.get_feature_names_out()

    # Typo correction
    corrected_words = []

    for word in processed_query.split():

        matches = get_close_matches(word, vocabulary, n=1, cutoff=0.7)

        if matches:
            corrected_words.append(matches[0])
        else:
            corrected_words.append(word)

    corrected_query = " ".join(corrected_words)

    # Convert query into vector
    user_vector = vectorizer.transform([corrected_query])

    # Compute similarity
    similarity_scores = cosine_similarity(user_vector, faq_vectors)

    # Best match
    best_match_index = similarity_scores.argmax()

    # Confidence score
    max_similarity = similarity_scores.max()

    # Response logic
    if max_similarity > 0.5:
        return faq_answers[best_match_index]
    else:
        return "Sorry, I don't understand the question."
```

```python
#Create Gradio Interface
interface = gr.Interface(
    fn=chatbot,
    inputs="text",
    outputs="text",
    title="FAQ AI Agent",
    description="Ask questions about AI, ML, Python, NLP and more."
)
```

```python
#Launch Chatbot
interface.launch()
```
