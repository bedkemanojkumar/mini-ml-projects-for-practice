```python

import pandas as pd
import matplotlib.pyplot as plt
import re

from langdetect import detect, detect_langs
from googletrans import Translator
```

```python
# Translator Object
translator = Translator()

# Language Dictionary
languages = {
    "en": "English",
    "te": "Telugu",
    "es": "Spanish",
    "fr": "French",
    "ta": "Tamil",
    "de": "German",
    "hi": "Hindi",
    "it": "Italian",
    "pt": "Portuguese",
    "ru": "Russian",
    "zh-cn": "Chinese",
    "so": "Somali"
}
```

```python
# Text Cleaning Function
def clean_text(text):

    # Convert to lowercase
    text = text.lower()

    # Remove URLs
    text = re.sub(r'http\\S+', '', text)

    # Remove email addresses
    text = re.sub(r'\\S+@\\S+', '', text)

    # Keep multilingual characters
    text = re.sub(
        r'[^A-Za-zÀ-ÿ\\u0B80-\\u0BFF\\u0C00-\\u0C7F\\u0900-\\u097F\\s]',
        '',
        text
    )

    # Remove extra spaces
    text = re.sub(r'\\s+', ' ', text).strip()

    return text
```

```python
# Preprocessing + Translation Function
def preprocess_and_translate(text):

    # Clean text
    cleaned = clean_text(text)

    # Translate to English
    translated = translator.translate(cleaned, dest='en')

    return cleaned, translated.text

print("Setup Completed Successfully")
```

```python
#Email Category Detection
def categorize_email(text):

    text = text.lower()

    if 'meeting' in text or 'schedule' in text:
        return 'Meeting Email'

    elif 'complaint' in text or 'issue' in text:
        return 'Complaint Email'

    elif 'offer' in text or 'discount' in text:
        return 'Promotional Email'

    elif 'support' in text or 'help' in text:
        return 'Support Email'

    else:
        return 'General Email'
```

```python
# STEP 10 — Confidence Visualization

import matplotlib.pyplot as plt
from langdetect import detect_langs

# Input
text = input("Enter Email/Text: ")

# Clean text
cleaned = clean_text(text)

# Get confidence scores
confidence = detect_langs(cleaned)

# Store data
languages_list = []
probabilities = []

# Extract language + probability
for item in confidence:

    # Example item -> te:0.99999
    parts = str(item).split(':')

    lang_code = parts[0]
    probability = float(parts[1])

    # Convert language code to full name
    lang_name = languages.get(lang_code, lang_code)

    languages_list.append(lang_name)
    probabilities.append(probability)

# Print confidence scores
print("\nConfidence Scores:\n")

for lang, prob in zip(languages_list, probabilities):

    print(f"{lang} --> {prob*100:.2f}%")

# Plot graph
plt.figure(figsize=(7,4))

plt.bar(languages_list, probabilities)

plt.xlabel("Languages")
plt.ylabel("Confidence Score")
plt.title("Language Detection Confidence Visualization")

plt.ylim(0,1)

plt.show()
```

```python



# Sample Dataset
sample_data = {
    'Email': [
        'Hello team we have a meeting tomorrow',
        'Hola necesitamos reunirnos mañana',
        'Bonjour comment allez vous',
        'వందనం మిత్రులారా',
        'Discount offer available today',
        'Please contact support team'
    ]
}

# Create DataFrame
df = pd.DataFrame(sample_data)

# Empty lists
detected_languages = []
translated_emails = []

# Process emails
for email in df['Email']:

    cleaned, translated = preprocess_and_translate(str(email))

    # Detect language
    detected = detect(cleaned)

    # Store results
    detected_languages.append(
        languages.get(detected, detected)
    )

    translated_emails.append(translated)

# Add new columns
df['Detected Language'] = detected_languages

df['Translated Email'] = translated_emails

# Final Output
print(df)
```

```python
#Accuracy Testing
test_data = {
    'Text': [
        'Hello friend',
        'Hola amigo',
        'Bonjour ami',
        'வணக்கம் நண்பரே'
    ],

    'Actual': [
        'English',
        'Spanish',
        'French',
        'Tamil'
    ]
}

accuracy_df = pd.DataFrame(test_data)

predictions = []

for text in accuracy_df['Text']:

    detected = detect(clean_text(text))

    predictions.append(languages.get(detected, detected))

accuracy_df['Predicted'] = predictions

correct = 0

for i in range(len(accuracy_df)):

    if accuracy_df['Actual'][i] == accuracy_df['Predicted'][i]:
        correct += 1

accuracy = (correct / len(accuracy_df)) * 100

print(accuracy_df)

print(f'\nModel Accuracy: {accuracy}%')
```
