```python
# Once again check percentage of outliers perm coloumn
def outlier_percentage(df, col):
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    return len(outliers) / len(df) * 100
```

```python
# Remove target column
num_cols = df.select_dtypes(include=['int64', 'float64']).columns # Re-initialize num_cols
num_cols = num_cols.drop('Churned')
for col in num_cols:
    print(col, ":", round(outlier_percentage(df, col), 2), "%")
```

```python
# remove outliers for numeric values
# Capping
for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df[col] = df[col].clip(lower, upper)
```

```python
import numpy as np

def regularized_loss(y_true, y_pred, weights, reg_alpha=0.5, reg_lambda=1.0):
    """
    y_true   : actual values
    y_pred   : predicted values
    weights  : model weights (simulated)
    reg_alpha: L1 regularization (alpha)
    reg_lambda: L2 regularization (lambda)
    """

    # Mean Squared Error (data loss)
    mse_loss = np.mean((y_true - y_pred) ** 2)

    # L1 Regularization
    l1_penalty = reg_alpha * np.sum(np.abs(weights))

    # L2 Regularization
    l2_penalty = reg_lambda * np.sum(weights ** 2)

    # Total Loss
    total_loss = mse_loss + l1_penalty + l2_penalty

    return total_loss
```
